## 1. Graphs

The PageRank corresponds to the stationary distribution of a random walk with restart probability $1-\alpha$. Unless otherwise specified, we take the default value $\alpha = 0.85$ and the restart probability distribution is uniform over the set of nodes.

## Linear graph

Consider a linear graph:

```python
n = 10
```

```python
dataset = linear_graph(n, True)
adjacency = dataset.adjacency
position = dataset.position
```

```python
image = visualize_graph(adjacency, position, names=np.arange(n))
SVG(image)
```

## To do

* What are the two best ranked nodes? Try with different values of $\alpha$ and interpret the results.
* What is the exact PageRank vector when $\alpha=1$ (no restarts)? Justify your answer.

#### **Answers**
- The best two ranked nodes are :
    - 1 and 8 for $\alpha = 0.85$. That happens because 1 and 8 are adjacent to the extremities, that means that if we are in 0 or 9,they have a higher chance of being reached since they are the only neighbors. Even if 0 and 9 are more difficult to reach, the restarting makes it possible.
    - 1 and 8 for $\alpha = 0.5$. Here we have the same answer.
    - 1 and 8 for $\alpha = 0.01$. Here we have the same answer and we notice that the closer we get to zero, the greater is the advantage of 1 and 8 score compared to other nodes.

- For $\alpha = 0.999$ and $ \alpha=1 $ we know that the $\pi$ is going to be proportional to the degree. The nodes with the same degree get same score and the ones with least degrees get lower score. We can even compute the scores by using the formula $\pi(t+1) = \pi(t)P$ which gives us $\pi = (1/18, 2/18,...,2/18,1/18)$.

```python
pagerank = PageRank(damping_factor=0.999, n_iter=100)
scores = pagerank.fit_predict(adjacency)
image = visualize_graph(adjacency, position, names=np.arange(n), scores=scores)
SVG(image)
```

```python
for alpha in [0.01,0.5,0.85,0.999]:
  pagerank = PageRank(damping_factor=alpha, n_iter=100)
  scores = pagerank.fit_predict(adjacency)
  print(f"{scores}\n")
```

#### **Answers**
For all values of $\alpha$ we see "Valjean" as a top ranked node. This happens because it is both the top degree and top weight node. Hence, regardless of the restart probability, it will have a high score.

For $\alpha = 0.001$, $\alpha = 0.5$ and $\alpha = 0.85$, we have "Myriel" at second, second and third position in the top10 rank for scores. This happens due to the restart scenario along with the fact that "Myriel" node is connected to a lot of leaves. Thus, if the restart goes to any of these leaves connected to "Myriel", the chances of "Myriel" being the next node is really high and that makes his rank higher.

Javert is also a interesting case. It does not appears as a top weight or degree node, but it is part of top10 scores for the values 0.5, 0.85 and 0.999 of $\alpha$. This happens due to its connections to several top degree nodes, which give him more chance of being visited.

The other top ranked nodes belong to the top weight or top degree list.

```python
pagerank = PageRank(damping_factor=0.001)

scores = pagerank.fit_predict(adjacency)

# scores in log scale appear more clearly
image = visualize_graph(adjacency, position, names, scores=np.log(scores), scale=3)
SVG(image)
```

```python
for alpha in [0.001,0.5,0.85,0.999]:
  pagerank = PageRank(damping_factor=alpha)
  scores = pagerank.fit_predict(adjacency)

  # list ten best ranked
  top_ranked = np.argsort(scores)[::-1][:10]

  print(f"ALPHA: {alpha}")
  print(names[top_ranked])

# list highest degrees
degrees = np.array((dataset.adjacency > 0).sum(axis=1)).flatten()
top_degrees = np.argsort(degrees)[::-1][:10]

# list highest weights
weights = np.array(dataset.adjacency.sum(axis=1)).flatten()
top_weights = np.argsort(weights)[::-1][:10]

print("TOP DEGREES")
print(names[top_degrees])
print("TOP WEIGHTS")
print(names[top_weights])
```

## 3. Bipartite graphs

## Cinema

```python
dataset = cinema
```

```python
biadjacency = dataset.biadjacency
movies = dataset.names_row
actors = dataset.names_col
```

## To do


* List the 5 closest actors and the 5 closest movies to **Catherine Deneuve**.

```python
# find Catherine Deneuve
target_actor = np.where(actors == "Catherine Deneuve")[0][0]

n_movies, n_actors = biadjacency.shape

adjacency = sparse.bmat([
    [None, biadjacency],
    [biadjacency.T, None]
], format="csr")

target_node = n_movies + target_actor

weights = {target_node: 1}

pagerank = PageRank(damping_factor=0.85)
scores = pagerank.fit_predict(adjacency, weights=weights)

movie_scores = scores[:n_movies]
actor_scores = scores[n_movies:]

top_movies = np.argsort(movie_scores)[::-1][:5]
# take top 6 because the first one is catherine herself
top_actors = np.argsort(actor_scores)[::-1][:6]

print("Closest movies:")
print(movies[top_movies])

print("Closest actors:")
print(actors[top_actors])
```

## 4. Directed graphs as bipartite graphs

Directed graphs can be represented as bipartite graphs by duplicating each node, one as source of edges and the other as destination of edges. The biadjacency matrix of this bipartite graph is simply the adjacency matrix of the original directed graph.

The PageRank scores obtained with the bipartite graph differ from those obtained with the directed graph: they correspond to the **forward-backward** random walk in the directed graph, edges being alternately followed in forward and backward directions.


## Wikipedia Vitals

```python
dataset = wikivitals

adjacency = dataset.adjacency
names = dataset.names
```

## To do

Do the same experiments as above and compare both rankings:

* List the 20 closest pages to **Pablo Picasso** in Wikipedia Vitals.
* List the 20 closest pages to both **Cat** and **Dog** in Wikipedia Vitals.

#### **Answers**

The bipartited approach changed the rank indeed. Now the rank is given by more connected nodes and not only highly weighted ones.

We also realize that the top ranked nodes here refer to more general terms as "Europe", instead of single countries (except France).

For Picasso, we note that the most different terms are the ones that represent unrelevant fields with respect to Picasso, such as Biology or Chemistry.

For Cat and Dog, we note the same logic, but with different fields being the more different ones, such as History or Philosophy.

```python
pagerank = PageRank()
bip_general_scores = pagerank.fit_predict(adjacency, force_bipartite=True)

# top ranks
bip_top_ranked = np.argsort(bip_general_scores)[::-1][:10]

print("TOP SCORES")
print(names[bip_top_ranked])
```

```python
#difference between this and non-bipartited version

diff_general = general_scores - bip_general_scores

top_diff_general = np.argsort(diff_general)[::-1][:10]

print("TOP DIFFERENCES")
print(names[top_diff_general])
```

```python
# find picasso
bip_picasso = np.where(np.char.find(np.char.lower(names.astype(str)), "pablo picasso") >= 0)[0]

weights = {i: 1 / len(bip_picasso) for i in bip_picasso}


#personalized page rank starting from picasso
pagerank = PageRank()

#personalized starting from picasso
bip_scores_picasso = pagerank.fit_predict(adjacency, weights=weights, force_bipartite=True)

bip_top_ranked = np.argsort(bip_scores_picasso)[::-1][:20]

print("TOP SCORES")
print(names[bip_top_ranked])
```

```python
# compute differences

diff_picasso = bip_general_scores - bip_scores_picasso

top_diff_picasso = np.argsort(diff_picasso)[::-1][:20]

print("TOP DIFFERENCES")
print(names[top_diff_picasso])
```

```python
# find cats and dogs indices
cat = np.where(names == "Cat")[0]
dog = np.where(names == "Dog")[0]
cat_dog = np.concatenate([cat, dog])


weights = {i: 1 / len(cat_dog) for i in cat_dog}


#personalized page rank starting from cat_dog
pagerank = PageRank()

#personalized starting from cat_dog
bip_scores_cat_dog = pagerank.fit_predict(adjacency, weights=weights, force_bipartite=True)

bip_top_ranked = np.argsort(bip_scores_cat_dog)[::-1][:20]

print("TOP SCORES")
print(names[bip_top_ranked])
```

```python
# compute differences

diff_cat_dog = bip_general_scores - bip_scores_cat_dog

top_diff_cat_dog = np.argsort(diff_cat_dog)[::-1][:20]

print("TOP DIFFERENCES")
print(names[top_diff_cat_dog])
```
