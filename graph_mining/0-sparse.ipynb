## To do

Consider the following matrix:

```python
n_row, n_col = 20, 4
X = sparse.csr_matrix(np.random.randint(3, size = (n_row, n_col)))
```

* Compute the vector of the Euclidean norm of each row.

```python
def euc_norm(X):
  norm = np.zeros((len(X.indptr)-1))
  for (i,_) in enumerate(X.indptr):
    if (i == (len(X.indptr)-1)): # break at last line
      break
    sum = 0
    for j in range(X.indptr[i],X.indptr[i+1]):
      sum += (X.data[j]**2)
    norm[i] = sum
    norm = np.sqrt(norm)
  return norm

print(euc_norm(X))
```

## To do

Consider the following matrix:

```python
n_row, n_col = 20, 4
X = sparse.csr_matrix(np.random.randint(2, size = (n_row, n_col)))
```

Using sparse diagonal matrices:
* Normalize this matrix so that each row sums to 1 (or to 0 if the whole row is zero).

```python
def normalize(X):
    X = X.astype(float)

    row_sums = np.array(X.sum(axis=1)).flatten()
    row_sums[row_sums == 0] = 1  # avoid division by zero

    D_inv = sparse.diags(1 / row_sums)

    return D_inv @ X
print(normalize(X))
```

## COO format

Another way to represent sparse matrices is the COO (COOrdinate) format. It is useful to load a matrix from a list of entries.

```python
row = [1, 4, 2]
col = [2, 0, 2]
data = [1, 2, 3]
```

```python
X_coo = sparse.coo_matrix((data, (row, col)), shape=(5, 5))
```

```python
X_coo
```

```python
X_coo.row
```

```python
X_coo.col
```

```python
X_coo.data
```

You can change the format:

```python
X_csr = X_coo.tocsr()
```

```python
X_csr
```

```python
X_csr.indices
```

```python
X_csr.indptr
```

```python
X_csr.data
```

```python
X_csr.tocoo()
```

You can directly load a CSR matrix from COO format:

```python
X_csr = sparse.csr_matrix((data, (row, col)), shape=(5, 5))
```

```python
X_csr
```
